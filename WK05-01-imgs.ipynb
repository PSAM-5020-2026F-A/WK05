```python
import matplotlib.pyplot as plt

from os import listdir
from PIL import Image as PImage

from image_utils import get_pixels

from Homework04_utils import Homework04Utils
```

```python
# TODO: List images

cloud_fnames = sorted([fn for fn in listdir("./data/image/clouds") if fn.endswith(".jpg")])
flower_fnames = sorted([fn for fn in listdir("./data/image/flowers") if fn.endswith(".png")])

print(len(cloud_fnames), len(flower_fnames))

# TODO: Separate into train/test
cloud_train = cloud_fnames[:250]
flower_train = flower_fnames[:250]

print(len(cloud_train), len(flower_train))
```

```python
# TODO: iterate through all images
#       get blue and/or green ratios

cloud_g_ratio = []
flower_g_ratio = []

cloud_b_ratio = []
flower_b_ratio = []

GREEN_REF = [0, 180, 0]
BLUE_REF  = [84, 141, 255]

# cloud images
for fn in cloud_train:
  img = PImage.open("./data/image/clouds/"+fn)
  pxs = get_pixels(img)
  g_ratio = Homework04Utils.color_ratio(pxs, GREEN_REF)
  b_ratio = Homework04Utils.color_ratio(pxs, BLUE_REF)
  cloud_g_ratio.append(g_ratio)
  cloud_b_ratio.append(b_ratio)

# flower images
for fn in flower_train:
  img = PImage.open("./data/image/flowers/"+fn).convert("RGB")
  pxs = get_pixels(img)
  g_ratio = Homework04Utils.color_ratio(pxs, GREEN_REF)
  b_ratio = Homework04Utils.color_ratio(pxs, BLUE_REF)
  flower_g_ratio.append(g_ratio)
  flower_b_ratio.append(b_ratio)

print(len(cloud_g_ratio), len(flower_g_ratio))
```

```python
# quick check
#   are clouds bluer ?
#   are flowers greener ?
print("green, cloud, flower")
print(sum(cloud_g_ratio)/len(cloud_g_ratio), sum(flower_g_ratio)/len(flower_g_ratio))

print("blue, cloud, flower")
print(sum(cloud_b_ratio)/len(cloud_b_ratio), sum(flower_b_ratio)/len(flower_b_ratio))
```

```python
# TODO: plot blue and/or green ratios for both classes of images

plt.scatter(len(cloud_g_ratio) * [0], cloud_g_ratio, color=(0,0.8,0.8), s=2, alpha=0.5)
plt.scatter(len(cloud_b_ratio) * [1], cloud_b_ratio, color=(0,0.2,0.8), s=2, alpha=0.5)

plt.scatter(len(flower_g_ratio) * [2], flower_g_ratio, color=(0,0.8,0.2), s=2, alpha=0.5)
plt.scatter(len(flower_b_ratio) * [3], flower_b_ratio, color=(0,0.8,0.8), s=2, alpha=0.5)

plt.show()
```

```python
plt.scatter(cloud_g_ratio, cloud_b_ratio, color=(0,0.2,0.8), s=2, alpha=0.5)
plt.scatter(flower_g_ratio, flower_b_ratio, color=(0,0.8,0.2), s=2, alpha=0.5)

plt.plot([0,0.55], [0.1,1], "-k")

plt.xlabel("green ratio")
plt.ylabel("blue ratio")
plt.show()
```

```python
def classifier(b_ratio, g_ratio):
  if b_ratio - g_ratio > 0.01:
    return "cloud"
  else:
    return "flower"

correct_cloud = 0
correct_flower = 0

for g_ratio, b_ratio in zip(cloud_g_ratio, cloud_b_ratio):
  if classifier(b_ratio, g_ratio) == "cloud":
    correct_cloud += 1

for g_ratio, b_ratio in zip(flower_g_ratio, flower_b_ratio):
  if classifier(b_ratio, g_ratio) == "flower":
    correct_flower += 1

print("cloud accuracy:", correct_cloud / len(cloud_g_ratio))
print("flower accuracy:", correct_flower / len(flower_g_ratio))
```
